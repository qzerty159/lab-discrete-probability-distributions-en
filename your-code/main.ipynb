## **Bernoulli's Distribution**

In a basket containing 100 fruits, there are 60 apples and 40 oranges. 

A fruit is picked randomly from the basket. 

1. **What is the probability that the fruit is apple?**
1. **What is the probability that it is an orange?**

````python
"""
Calculate:
p = probability that the fruit is an apple 
q = probability that the fruit is an orange
"""
````

### Probability the fruit is an apple

There are 60 apples out of 100 fruits.

$$ P(\text{apple}) = \frac{60}{100} = 0.6 $$

````python
p_apple = 60/100
print (f"The probability that the fruit is apple is {round(p_apple, 3)*100} %")
````

### Probability the fruit is an orange

There are 40 oranges out of 100 fruits. 
$$ Q(\text{orange}) = \frac{40}{100} = 0.4 $$

````python
q_orange = 40/100

print (f"The probability that the fruit is orange is {round(q_orange, 3)*100} %")
````

### Final Rounded Answers

* Apple: 0.600
* Orange: 0.400

Now we take a random sample of 20 fruits from the basket. After each fruit is taken, a new fruit of the same type is replaced in the basket. Therefore, every time we are taking 1 fruit from 100 fruits. 

1. **What is the probability that the first 5 fruits are all apples?**

1. **What is the probability that the first 5 fruits are all apples and the next 15 fruits are all oranges?**

You can include the `p` and `q` probabilities you previous calculated in your solution.

Because each fruit is replaced after drawing, every pick is an independent Bernoulli trial with:

* ( p = 0.6 ) = probability of apple
* ( q = 0.4 ) = probability of orange

### Probability the first 5 fruits are all apples

Since each draw is independent:
$$ P(\text{5 apples}) = p^5 = 0.6^5 $$

````python
p = 0.6
prob_5_apples = p**5
print(f"The probability the first 5 fruits are all apples is {round(prob_5_apples, 3)*100} %")
````

### Probability first 5 fruits are apples AND next 15 are oranges

Independence again allows multiplication:
$$ P = p^5 \times q^{15} $$

Where:
* ( p = 0.6 )
* ( q = 0.4 )

## Binomial Distribution

Similar to the above problem, this time again we randomly sample 20 fruits.

**What is the probability that this time the sample contains 5 apples and 15 oranges?**

Please note that this time the order of the fruits being picked does not matter.

The order does not matter, so this is a binomial distribution problem.

We are drawing 20 fruits with replacement, so each draw is an independent trial with:

* Probability of apple: p = 0.6
* Probability of orange: q = 0.4

We want exactly 5 apples (successes) out of 20 trials.

This is:

$$ P(X = 5) = \binom{20}{5} p^5 q^{15} $$

````python
from scipy.stats import binom

n = 20
p = 0.6

prob_5_apples = binom.pmf(5, n, p)
print(f"The probability that this time the sample contains 5 apples and 15 oranges is {round(prob_5_apples, 3)*100} %")
````

In the lesson we also discussed CDF (cumulative distribution function). In the context of the above problem, **what is the probability that less than 5 fruits picked are apples?**

Write the formula and calculate the probability below.

### Python Calculation (rounded to three decimals)

````python
from scipy.stats import binom

n = 20
p = 0.6

prob_less_than_5 = binom.cdf(4, n, p)
print(f"The probability that less than 5 fruits picked are apples is {round(prob_less_than_5, 3)*100} %")
````

The probability that less than 5 fruits are apples means:

$$ P(X < 5) = P(X = 0) + P(X = 1) + P(X = 2) + P(X = 3) + P(X = 4) $$

Since this is a binomial distribution with:

* ( n = 20 )
* ( p = 0.6 )

the cumulative probability is:

$$ P(X < 5) = \text{BinomCDF}(4; n=20, p=0.6) $$

This is exactly the CDF evaluated at 4.

**Plot the Probability Mass Function (PMF) for the above binomial distribution.**

*Hint: The range of x in a binomial distribution is from `0` to `n`.*

````python
n = 20
p = 0.6

# Range of possible values: 0 to n
x = np.arange(0, n + 1)

# PMF values
pmf_values = binom.pmf(x, n, p)

# Plot
plt.figure(figsize=(8, 5))
plt.stem(x, pmf_values, basefmt=" ")
plt.xlabel("Number of Apples (k)")
plt.ylabel("Probability Mass Function (PMF)")
plt.title("Binomial PMF (n=20, p=0.6)")
plt.grid(True)
plt.show()
````

## Poisson Distribution

In this challenge you are required to use the Math library in python to solve the problems.

In the UEFA champions league tournament, the average number of goals scored in a match is 2.3. 

**What is the probability that the number of goals scored in a match is 5?**

*Hint: Use the exp() in the math library*

And to calculate the factorial, you can use the following code:

```python
import math
math.factorial(n) # n is the number you want to find the factorial of
```

````python
import math

lam = 2.3
k = 5

prob_5_goals = math.exp(-lam) * (lam**k) / math.factorial(k)

print(f"The probability that the number of goals scored in a match is 5 : {round(prob_5_goals, 3)*100} % \n\n")
print("This means there is about an 8.5% chance that a Champions League match ends with exactly 5 goals when the average is 2.3.")
````

### Poisson PMF Formula

For a Poisson distribution with rate ( \lambda ):
$$ P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!} $$

Here:

* $ \lambda = 2.3 $ (average goals per match)
* $ k = 5 $ (goals you want the probability for)

---
---

**Draw a poisson probability distribution plot for the number of goals to be between 0 to 10.**

````python
import math
import matplotlib.pyplot as plt

lam = 2.3
x = list(range(0, 11))

pmf = [math.exp(-lam) * lam**k / math.factorial(k) for k in x]

plt.figure(figsize=(8,5))
plt.stem(x, pmf, basefmt=" ")
plt.xlabel('Number of Goals (k)')
plt.ylabel('Probability Mass Function (PMF)')
plt.title('Poisson PMF for Goals (λ = 2.3)')
plt.grid(True)
plt.show()
````
